### Emperical Rule

- 68% of data available between u-1*sd to u+1*sd
- step1: calculate mean of wage: mean
- step2: calculate sd of wage: sd
- step3: lb=mean-1*sd
- step4: ub=mean+1*sd
- step5: con1: wage_data>lb
- step6: con2=wage_data<ub
- step7:con=con1 &con2
- step8: np.sum(con)

```python
wage_data=visa_data['prevailing_wage']
mean=wage_data.mean()
st=np.std(wage_data)
lb=mean-1*st
ub=mean+1*st
con1=wage_data>lb
con2=wage_data<ub
con=con1 & con2
print(np.sum(con)) 
print(68*25480/100)
np.sum(con)==68*25480/100
```

**how many outliers calculate**

- step1: cal q1
- 2: calq2
- 3: cal q3
- 4: IQR=q3-q1
- 5: lb=q1-1.5*iqr
- 6: ub=q3-1.5*iqr
- 7: con1: wage_data<lb
- 8: con2: wage_data>ub
- 9: con=con1 & con2
- 10: np.sum()


```python
Q1=np.percentile(wage_data,25)
Q2=np.percentile(wage_data,50)
Q3=np.percentile(wage_data,75)
IQR=Q3-Q1
lb=Q1-1.5*IQR
ub=Q3+1.5*IQR
con1=wage_data<lb
con2=wage_data>up
con=con1 | con2
np.sum(con)
```

## **capping**
- we can fill the outliers which is less than 25p with 25p value
- we can fill the outlier which is greater than 75p with 75p value

```python
wage_data=visa_data['prevailing_wage']
q1=np.percentile(wage_data,25)
q2=np.percentile(wage_data,50)
q3=np.percentile(wage_data,75)
iqr=q3-q1
lb=q1-1.5*iqr
ub=q3+1.5*iqr
visa_data['wage_clip']=wage_data.clip(lb,ub)
```

```python
wage_clip=visa_data['wage_clip']
plt.figure(figsize=(15,4))
plt.subplot(1,3,1).boxplot(wage_median,vert=False)
plt.subplot(1,3,2).boxplot(wage_data,vert=False)
plt.subplot(1,3,3).boxplot(wage_clip,vert=False)
plt.show()
```
