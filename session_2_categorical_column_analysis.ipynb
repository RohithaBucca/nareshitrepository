```python
cont=visa['continent'].unique()
freq=[]
for i in cont:
    count=visa['continent']==i    #asia
    freq.append(np.sum(count))

df=pd.DataFrame(zip(cont,freq),columns=['Continent_class','frequency'])
df
```
