```python
visa.dtypes
# idea
# exr=tract keys and values
# apply the for loop on values
#for i,j in zip(values,keys):
# if i=='object':
# j
```

```python
values=visa.dtypes.values.tolist()
keys=visa.dtypes.keys().tolist()
```

```python
cat=[]
num=[]
for i,j in zip(values,keys):
    if i=='object':
        cat.append(j)
    else:
        num.append(j)
cat
```

```python
stri=['h','e','l','l','o']
vow=['a','e','i','o','u']
for i in stri:
    for j in vow:
        if i==j:
            print(i)            
```

```python
stri=['h','e','l','l','o']
vow=['a','e','i','o','u']

for i,j in zip(stri,vow):
    if i==j:
        print(i)
```
