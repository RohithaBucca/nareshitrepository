## pd.get_dummies  ( also called One-Hot Encoder)

```python
visa_data['case_status']
```

- get dummies is a way to represent the categorical columns into numerical
- first it will check what are the labels in a column
- for ex: case_status has 2 labels
      - denied and certified
- so it will create two new columns based on labels
    - col1: case_status_Denied
    - col2: case_status_certified
- new columns are **independent** to eachother (90Degree)
- also called as **One-Hot encoder**
- some of ML algorithms expect that columns should be **independent** each other



case_status | case_status_certified | case_status_denied
 denied     |          0            |         1
 certified  |          1            |         0
 certified  |          1            |         0
