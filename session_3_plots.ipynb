### Pie Chart

- autopct= '%a.bf%%'
- first % is the formatting
- a  : how much width
- b  : after point how many decimals
- %% : % symbol
- %1.1f%% : 4.2%   40.2%   400.2%
- %2.2f%% : 40.23%   400.78%

```python
keys=visa_data['continent'].value_counts().keys()
value=visa_data['continent'].value_counts().values
keys,value
```

```python
colors=['plum','lavenderblush','pink','springgreen','yellow','red']
plt.pie(value,
        labels=keys,
        explode=[0.03 for i in range(6)],
        colors=colors,
        autopct='%1.2f%%',
        radius=0.8,
       startangle=180,
       )
plt.savefig('continent_pie.jpg')
plt.show()
```

```python
# case3: using for loop

cols=['continent','case_status']
plt.figure(figsize=(18,3))
for i in range(len(cols)):
    keys=visa_data[cols[i]].value_counts().keys()
    value=visa_data[cols[i]].value_counts().values
    plt.subplot(1,2,i+1).bar(keys,value)
```
